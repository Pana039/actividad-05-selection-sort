```python
def calc_EST(aaa):
    x1 = []
    for v in aaa:
        if v is not None and v >= 0:
            x1.append(float(v))
    n = len(x1)
    if n == 0:
        return None
    m1 = sum(x1) / n
    v1 = sum((x - m1) ** 2 for x in x1) / n
    ds_1 = v1 ** 0.5
    return {"m": m1, "v": v1, "d": ds_1}

datos_XYZ = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]
res = calc_EST(datos_XYZ)
print("Resultado:", res)
```

```python
def filtrar_lecturas_validas(datos_entrada):
    lecturas_validas = []

    for lectura in datos_entrada:
        if lectura is not None and lectura >= 0:
            lecturas_validas.append(float(lectura))

    return lecturas_validas


def calcular_estadisticas_descriptivas(lecturas_validas):
    cantidad = len(lecturas_validas)

    if cantidad == 0:
        return None

    media = sum(lecturas_validas) / cantidad

    varianza = sum(
        (lectura - media) ** 2 for lectura in lecturas_validas
    ) / cantidad

    desviacion_estandar = varianza ** 0.5

    return {
        "m": media,
        "v": varianza,
        "d": desviacion_estandar,
    }


def procesar_analisis_estadistico(datos_crudos):
    lecturas_validas = filtrar_lecturas_validas(datos_crudos)

    return calcular_estadisticas_descriptivas(lecturas_validas)


# Dataset de prueba original
datos_crudos = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]

resultado = procesar_analisis_estadistico(datos_crudos)

print("Resultado:", resultado)
```
