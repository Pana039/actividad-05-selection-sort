```python
x = [34,18,42,12,29]
i = 0
```

```python
for i in range(len(x)):
  minimo = i
  for j in range(i+1,len(x)):
    if x[j] < x[minimo]:
      minimo = j
  if minimo != i:
    x[i],x[minimo] = x[minimo],x[i]

print (x)
```
